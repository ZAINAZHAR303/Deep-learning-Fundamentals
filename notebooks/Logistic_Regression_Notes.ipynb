# Logistic Regression with a Neural Network Mindset

## Core Concepts & Definitions
* **Logistic Regression as a Neural Network:** While traditionally viewed as a statistical method, logistic regression can be represented as a 1-layer neural network (a single neuron). It takes input features, applies a linear transformation (weights and bias), and passes the result through a non-linear activation function to output a probability.
* **Forward Propagation:** The process of pushing input data ($X$) through the network to generate a prediction ($A$) and calculate the current error or **Cost** ($J$). 
* **Backward Propagation (Backprop):** The process of calculating the gradients (derivatives) of the cost function with respect to the model's parameters ($w$ and $b$). These gradients (`dw`, `db`) indicate the direction of the steepest slope of the error surface.
* **Gradient Descent:** The optimization algorithm that updates the parameters by taking small steps (determined by the `learning_rate`, $\alpha$) in the opposite direction of the gradients to minimize the cost.
* **Sigmoid Activation Function:** A mathematical function that maps any real-valued number into a range between 0 and 1. It is essential for binary classification to output a probability.
* **Broadcasting:** A NumPy feature that automatically expands smaller arrays to match the dimensions of larger arrays during arithmetic operations.

## Key Learnings & General Architecture
Building a neural learning algorithm consistently follows this pipeline:
1. **Define Model Structure:** Determine input dimensions.
2. **Initialize Parameters:** Set initial weights $w$ to zeros and bias $b$ to zero.
3. **Optimization Loop:**
    * Compute current loss (Forward Prop).
    * Compute gradients (Backward Prop).
    * Update parameters using gradient descent.
4. **Prediction:** Use the final optimized parameters to evaluate accuracy on unseen test data.
